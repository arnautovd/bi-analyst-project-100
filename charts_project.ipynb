# Анализ визитов и регистраций

Загрузка данных о визитах и регистрациях из API и предварительный анализ.

```python
import os

import pandas as pd
import requests
from dotenv import load_dotenv

load_dotenv()  # загружаются переменные из файла .env, если он есть

API_URL = os.getenv('API_URL')
DATE_BEGIN = os.getenv('DATE_BEGIN')
DATE_END = os.getenv('DATE_END')

missing = [name for name, value in
           {'API_URL': API_URL, 'DATE_BEGIN': DATE_BEGIN, 'DATE_END': DATE_END}.items()
           if not value]
if missing:
    raise RuntimeError(f'Не заданы переменные окружения: {", ".join(missing)}. См. .env.example')

print(f'API: {API_URL}, период: {DATE_BEGIN} -> {DATE_END}')
```

## Загрузка данных из API

Маршруты `/visits` и `/registrations` принимают `begin` и `end` в формате `YYYY-MM-DD`. Дата `end` в выборку не входит.

```python
def fetch_data(route: str, begin: str, end: str) -> pd.DataFrame:
    """Запрашивает данные из API за период [begin, end) и возвращает DataFrame."""
    response = requests.get(
        f'{API_URL}/{route}',
        params={'begin': begin, 'end': end},
        timeout=120,
    )
    response.raise_for_status()
    df = pd.DataFrame(response.json())
    df['datetime'] = pd.to_datetime(df['datetime'])
    return df


visits = fetch_data('visits', DATE_BEGIN, DATE_END)
registrations = fetch_data('registrations', DATE_BEGIN, DATE_END)
print(f'Визиты: {len(visits)}, регистрации: {len(registrations)}')
```

## Конверсия визитов в регистрации

1. Убираем визиты ботов: поисковые и SEO-боты определяются по слову `bot` в User-Agent.
2. Пользователь мог заходить несколько раз до регистрации, поэтому для каждого `visit_id` берём только последний визит.
3. Группируем визиты и регистрации по дате и платформе, объединяем и считаем конверсию в процентах.

```python
is_bot = visits['user_agent'].str.contains('bot', case=False)
print(f'Визитов ботов: {is_bot.sum()}')

last_visits = (
    visits[~is_bot]
    .sort_values('datetime')
    .drop_duplicates(subset='visit_id', keep='last')
)
print(f'Визитов без ботов, последний визит на visit_id: {len(last_visits)}')
```

```python
visits_by_day = (
    last_visits
    .assign(date_group=last_visits['datetime'].dt.normalize())
    .groupby(['date_group', 'platform'])
    .size()
    .reset_index(name='visits')
)

registrations_by_day = (
    registrations
    .assign(date_group=registrations['datetime'].dt.normalize())
    .groupby(['date_group', 'platform'])
    .size()
    .reset_index(name='registrations')
)
```

```python
conversion = visits_by_day.merge(
    registrations_by_day, on=['date_group', 'platform'], how='outer'
)
conversion[['visits', 'registrations']] = (
    conversion[['visits', 'registrations']].fillna(0).astype(int)
)
conversion['conversion'] = conversion['registrations'] / conversion['visits'] * 100
conversion = conversion.sort_values(['date_group', 'platform']).reset_index(drop=True)
conversion.head(10)
```

```python
conversion.describe()
```

```python
# Дата сохраняется как Unix-время в миллисекундах (формат из примера задания)
epoch_ms = (conversion['date_group'] - pd.Timestamp(0)) // pd.Timedelta(milliseconds=1)
conversion.assign(date_group=epoch_ms).to_json('./conversion.json')
```

## Рекламные кампании

Читаем `./ads.csv` с затратами на рекламу и присоединяем к визитам и регистрациям по дате. Для дней без рекламы `cost = 0`, `utm_campaign = 'none'`.

```python
ads = pd.read_csv('./ads.csv', parse_dates=['date'])
ads.info()
ads.head()
```

```python
ads.describe(include='all')
```

```python
ads_by_day = (
    ads
    .assign(date_group=ads['date'].dt.normalize())
    .groupby('date_group')
    .agg(cost=('cost', 'sum'), utm_campaign=('utm_campaign', 'first'))
    .reset_index()
)

totals_by_day = (
    conversion
    .groupby('date_group')[['visits', 'registrations']]
    .sum()
    .reset_index()
)

ads_conversion = totals_by_day.merge(ads_by_day, on='date_group', how='left')
ads_conversion['cost'] = ads_conversion['cost'].fillna(0).astype(int)
ads_conversion['utm_campaign'] = ads_conversion['utm_campaign'].fillna('none')
ads_conversion = ads_conversion.sort_values('date_group').reset_index(drop=True)
ads_conversion.head(10)
```

```python
ads_conversion.groupby('utm_campaign').agg(
    start=('date_group', 'min'),
    end=('date_group', 'max'),
    days=('date_group', 'count'),
    visits=('visits', 'sum'),
    registrations=('registrations', 'sum'),
    cost=('cost', 'sum'),
).sort_values('start')
```

```python
epoch_ms = (ads_conversion['date_group'] - pd.Timestamp(0)) // pd.Timedelta(milliseconds=1)
ads_conversion.assign(date_group=epoch_ms).to_json('./ads.json')
```

## Визуализация

Графики сохраняются в PNG в директорию `./charts`. Все графики содержат числовые подписи:

- динамика визитов, регистраций и конверсии — **по неделям**, значение подписано у каждого столбца или точки;
- графики с рекламными кампаниями — **по дням**, над каждой кампанией и паузой подписано среднее за день (затраты — сумма).

```python
import matplotlib

matplotlib.use('Agg')  # чтобы ноутбук выполнялся без дисплея через jupyter execute
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Image, display

%matplotlib inline

CHARTS_DIR = './charts'
os.makedirs(CHARTS_DIR, exist_ok=True)

sns.set_theme(style='whitegrid')
FIGSIZE = (14, 7)
LABEL_FONTSIZE = 8
PLATFORMS = ['web', 'android', 'ios']
PLATFORM_COLORS = dict(zip(PLATFORMS, sns.color_palette('Set2', len(PLATFORMS))))
REGISTRATION_TYPES = ['email', 'google', 'apple', 'yandex']
TYPE_COLORS = dict(zip(REGISTRATION_TYPES, sns.color_palette('Set2', len(REGISTRATION_TYPES))))

campaigns = (
    ads_conversion.loc[ads_conversion['utm_campaign'] != 'none']
    .groupby('utm_campaign')['date_group'].min()
    .sort_values()
    .index
    .tolist()
)
CAMPAIGN_COLORS = dict(zip(campaigns, sns.color_palette('tab10', len(campaigns))))
CAMPAIGN_COLORS['none'] = 'lightgray'


def fmt_int(value):
    """1234567 -> '1 234 567'."""
    return f'{value:,.0f}'.replace(',', ' ')


def save_chart(fig, name):
    """Сохраняет график в ./charts/<name>.png и закрывает фигуру."""
    path = os.path.join(CHARTS_DIR, f'{name}.png')
    fig.tight_layout()
    fig.savefig(path, dpi=150)
    plt.close(fig)
    display(Image(path))


# --- Недельные данные: 184 дня не подписать, 27 недель — можно ---

def add_week(df):
    """Добавляет столбец week — понедельник недели, к которой относится дата."""
    return df.assign(week=df['date_group'].dt.to_period('W-SUN').dt.start_time)


week_days = add_week(ads_conversion).groupby('week')['date_group'].agg(['min', 'count'])
WEEKS = week_days.index
# Неделя подписывается первой датой с данными; неполные недели помечаются звёздочкой
WEEK_LABELS = [
    first_day.strftime('%d.%m') + ('*' if days < 7 else '')
    for first_day, days in zip(week_days['min'], week_days['count'])
]
PARTIAL_WEEK_NOTE = '* неполная неделя (начало или конец периода)'


def format_week_axis(ax):
    ax.set_xticks(range(len(WEEKS)))
    ax.set_xticklabels(WEEK_LABELS, rotation=45)
    ax.set_xlabel(f'неделя (с понедельника)    {PARTIAL_WEEK_NOTE}')


def weekly_bars(ax, values, color='skyblue', fmt=fmt_int):
    """Столбцы по неделям с подписью значения над каждым столбцом."""
    values = values.reindex(WEEKS, fill_value=0)
    bars = ax.bar(range(len(values)), values.values, color=color)
    ax.bar_label(bars, labels=[fmt(v) for v in values.values], fontsize=LABEL_FONTSIZE, padding=2)
    ax.margins(y=0.15)
    format_week_axis(ax)


def weekly_line(ax, values, color, fmt=lambda v: f'{v:.1f}', label=None):
    """Линия по неделям с подписью значения у каждой точки."""
    values = values.reindex(WEEKS)
    x = range(len(values))
    ax.plot(x, values.values, marker='o', color=color, label=label)
    for xi, v in zip(x, values.values):
        ax.annotate(fmt(v), (xi, v), textcoords='offset points', xytext=(0, 6),
                    ha='center', fontsize=LABEL_FONTSIZE)
    ax.margins(y=0.2)
    format_week_axis(ax)


# --- Дневные данные с выделением кампаний ---

def format_date_axis(ax):
    ax.xaxis.set_major_locator(mdates.WeekdayLocator(byweekday=mdates.MO, interval=2))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%d.%m'))
    ax.tick_params(axis='x', rotation=45)
    ax.set_xlim(
        ads_conversion['date_group'].min() - pd.Timedelta(days=1),
        ads_conversion['date_group'].max() + pd.Timedelta(days=1),
    )


# Отрезки подряд идущих дней с одной кампанией (или паузой)
periods = (
    ads_conversion
    .assign(period=(ads_conversion['utm_campaign'] != ads_conversion['utm_campaign'].shift()).cumsum())
    .groupby('period')
    .agg(utm_campaign=('utm_campaign', 'first'), start=('date_group', 'min'), end=('date_group', 'max'),
         visits=('visits', 'mean'), registrations=('registrations', 'mean'),
         cost=('cost', 'mean'), cost_total=('cost', 'sum'))
)


def bars_by_campaign(ax, column, label_func):
    """Дневные столбцы, окрашенные по кампании, и подпись над каждым отрезком."""
    for campaign, color in CAMPAIGN_COLORS.items():
        part = ads_conversion[ads_conversion['utm_campaign'] == campaign]
        ax.bar(part['date_group'], part[column], width=0.8, color=color, label=campaign)
    top = ads_conversion[column].max()
    for _, period in periods.iterrows():
        start = period['start'] - pd.Timedelta(hours=12)
        end = period['end'] + pd.Timedelta(hours=12)
        mean = period[column]
        ax.hlines(mean, start, end, colors='black', linestyles='--', linewidth=1)
        label = label_func(period)
        if not label:
            continue
        ax.text(start + (end - start) / 2, top * 1.03, label,
                ha='center', va='bottom', fontsize=LABEL_FONTSIZE + 1, fontweight='bold')
    ax.set_ylim(0, top * 1.35)
    ax.legend(title='utm_campaign (пунктир — среднее за день)', loc='upper right', ncol=3, fontsize=9)
    format_date_axis(ax)
```

### Итоговые визиты

```python
weekly = add_week(ads_conversion).groupby('week')[['visits', 'registrations', 'cost']].sum()

fig, ax = plt.subplots(figsize=FIGSIZE)
weekly_bars(ax, weekly['visits'])
ax.set(title='Итоговые визиты по неделям', ylabel='visits')
save_chart(fig, 'total_visits')
```

У платформ разный масштаб (web на порядок больше), поэтому у каждой свой график.

```python
weekly_by_platform = (
    add_week(conversion)
    .groupby(['week', 'platform'])[['visits', 'registrations']]
    .sum()
    .unstack('platform')
)


def platform_subplots(column, title):
    fig, axes = plt.subplots(len(PLATFORMS), 1, figsize=(14, 12), sharex=True)
    for ax, platform in zip(axes, PLATFORMS):
        values = weekly_by_platform[(column, platform)]
        weekly_bars(ax, values, color=PLATFORM_COLORS[platform])
        ax.set(ylabel=column, xlabel='')
        ax.set_title(f'{platform}: {fmt_int(values.sum())} за период', loc='left', fontsize=11)
    format_week_axis(axes[-1])
    fig.suptitle(title)
    return fig


save_chart(platform_subplots('visits', 'Визиты по платформам по неделям'), 'visits_by_platform')
```

### Конверсия по каждой платформе

Конверсия web на порядок ниже, чем у приложений, поэтому у каждой платформы свой график и своя шкала.

```python
fig, axes = plt.subplots(len(PLATFORMS), 1, figsize=(14, 12), sharex=True)
for ax, platform in zip(axes, PLATFORMS):
    visits_week = weekly_by_platform[('visits', platform)]
    registrations_week = weekly_by_platform[('registrations', platform)]
    weekly_line(ax, registrations_week / visits_week * 100, color=PLATFORM_COLORS[platform], label=platform)
    mean = registrations_week.sum() / visits_week.sum() * 100
    ax.axhline(mean, color='black', linestyle='--', linewidth=1, label=f'за период: {mean:.1f}%')
    ax.set(ylabel='conversion, %', xlabel='')
    ax.legend(loc='lower left')
format_week_axis(axes[-1])
fig.suptitle('Конверсия визитов в регистрации по платформам, по неделям, %')
save_chart(fig, 'conversion_by_platform')
```

### Регистрации по способам регистрации

Доля каждого способа по неделям. Если бы какой-то способ регистрации сломался, его доля упала бы до нуля.

```python
registrations_by_type = (
    add_week(registrations.assign(date_group=registrations['datetime'].dt.normalize()))
    .pivot_table(index='week', columns='registration_type', values='user_id',
                 aggfunc='count', fill_value=0)
    .reindex(WEEKS, fill_value=0)
)
type_share = registrations_by_type[REGISTRATION_TYPES].div(registrations_by_type.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots(figsize=FIGSIZE)
bottom = pd.Series(0.0, index=type_share.index)
for reg_type in REGISTRATION_TYPES:
    bars = ax.bar(range(len(type_share)), type_share[reg_type], bottom=bottom,
                  color=TYPE_COLORS[reg_type], label=reg_type)
    ax.bar_label(bars, labels=[f'{v:.0f}' for v in type_share[reg_type]],
                 label_type='center', fontsize=LABEL_FONTSIZE)
    bottom += type_share[reg_type]
ax.set_ylim(0, 100)
ax.legend(title='registration_type', loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=4)
ax.set(title='Доля способов регистрации по неделям, %', ylabel='% регистраций за неделю')
format_week_axis(ax)
save_chart(fig, 'registrations_by_type')
```

## Эффективность рекламных кампаний

Базовый уровень — средние визиты и регистрации в день без рекламы. Прирост кампании — сколько визитов и регистраций в день она дала сверх базового уровня. Стоимость дополнительной регистрации = затраты в день / прирост регистраций в день.

```python
baseline = ads_conversion.loc[ads_conversion['utm_campaign'] == 'none', ['visits', 'registrations']].mean()

campaign_summary = (
    ads_conversion
    .groupby('utm_campaign')
    .agg(
        start=('date_group', 'min'),
        end=('date_group', 'max'),
        days=('date_group', 'count'),
        cost=('cost', 'sum'),
        visits_per_day=('visits', 'mean'),
        registrations_per_day=('registrations', 'mean'),
        cost_per_day=('cost', 'mean'),
        visits_total=('visits', 'sum'),
        registrations_total=('registrations', 'sum'),
    )
    .sort_values('start')
)
campaign_summary['conversion'] = (
    campaign_summary['registrations_total'] / campaign_summary['visits_total'] * 100
)
campaign_summary['extra_visits_per_day'] = campaign_summary['visits_per_day'] - baseline['visits']
campaign_summary['extra_registrations_per_day'] = (
    campaign_summary['registrations_per_day'] - baseline['registrations']
)
campaign_summary['cost_per_extra_registration'] = (
    campaign_summary['cost_per_day'] / campaign_summary['extra_registrations_per_day']
).where(campaign_summary.index != 'none')

numeric_columns = campaign_summary.select_dtypes('number').columns
campaign_summary[numeric_columns] = campaign_summary[numeric_columns].round(1)
campaign_summary.drop(columns=['visits_total', 'registrations_total'])
```

```python
conversion_by_campaign = (
    conversion
    .merge(ads_conversion[['date_group', 'utm_campaign']], on='date_group')
    .groupby(['utm_campaign', 'platform'])[['visits', 'registrations']]
    .sum()
)
conversion_by_campaign['conversion'] = (
    conversion_by_campaign['registrations'] / conversion_by_campaign['visits'] * 100
)
conversion_by_campaign['conversion'].unstack().loc[campaign_summary.index].round(1)
```

### Эффективность кампаний

```python
efficiency = campaign_summary.drop(index='none').sort_values('cost_per_extra_registration')

fig, axes = plt.subplots(1, 2, figsize=FIGSIZE, sharey=True)
colors = [CAMPAIGN_COLORS[c] for c in efficiency.index]

bars = axes[0].barh(efficiency.index, efficiency['extra_registrations_per_day'], color=colors)
axes[0].bar_label(bars, labels=[f'+{v:.1f}' for v in efficiency['extra_registrations_per_day']], padding=3)
axes[0].set(title='Дополнительных регистраций в день\n(к уровню дней без рекламы)', xlabel='registrations / day')

bars = axes[1].barh(efficiency.index, efficiency['cost_per_extra_registration'], color=colors)
axes[1].bar_label(bars, labels=[f'{v:.1f}' for v in efficiency['cost_per_extra_registration']], padding=3)
axes[1].set(title='Цена одной дополнительной регистрации\n(затраты в день / доп. регистраций в день)', xlabel='cost')

for ax in axes:
    ax.margins(x=0.15)
axes[0].invert_yaxis()
save_chart(fig, 'campaign_efficiency')
```
