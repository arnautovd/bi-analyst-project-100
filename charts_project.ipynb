# Анализ визитов и регистраций

Загрузка данных о визитах и регистрациях из API и предварительный анализ.

```python
import os

import pandas as pd
import requests
from dotenv import load_dotenv

load_dotenv()  # загружаются переменные из файла .env, если он есть

API_URL = os.getenv('API_URL')
DATE_BEGIN = os.getenv('DATE_BEGIN')
DATE_END = os.getenv('DATE_END')

missing = [name for name, value in
           {'API_URL': API_URL, 'DATE_BEGIN': DATE_BEGIN, 'DATE_END': DATE_END}.items()
           if not value]
if missing:
    raise RuntimeError(f'Не заданы переменные окружения: {", ".join(missing)}. См. .env.example')

print(f'API: {API_URL}, период: {DATE_BEGIN} -> {DATE_END}')
```

## Загрузка данных из API

Маршруты `/visits` и `/registrations` принимают `begin` и `end` в формате `YYYY-MM-DD`. Дата `end` в выборку не входит.

```python
def fetch_data(route: str, begin: str, end: str) -> pd.DataFrame:
    """Запрашивает данные из API за период [begin, end) и возвращает DataFrame."""
    response = requests.get(
        f'{API_URL}/{route}',
        params={'begin': begin, 'end': end},
        timeout=120,
    )
    response.raise_for_status()
    df = pd.DataFrame(response.json())
    df['datetime'] = pd.to_datetime(df['datetime'])
    return df


visits = fetch_data('visits', DATE_BEGIN, DATE_END)
registrations = fetch_data('registrations', DATE_BEGIN, DATE_END)
print(f'Визиты: {len(visits)}, регистрации: {len(registrations)}')
```

## Конверсия визитов в регистрации

1. Убираем визиты ботов: поисковые и SEO-боты определяются по слову `bot` в User-Agent.
2. Пользователь мог заходить несколько раз до регистрации, поэтому для каждого `visit_id` берём только последний визит.
3. Группируем визиты и регистрации по дате и платформе, объединяем и считаем конверсию в процентах.

```python
is_bot = visits['user_agent'].str.contains('bot', case=False)
print(f'Визитов ботов: {is_bot.sum()}')

last_visits = (
    visits[~is_bot]
    .sort_values('datetime')
    .drop_duplicates(subset='visit_id', keep='last')
)
print(f'Визитов без ботов, последний визит на visit_id: {len(last_visits)}')
```

```python
visits_by_day = (
    last_visits
    .assign(date_group=last_visits['datetime'].dt.normalize())
    .groupby(['date_group', 'platform'])
    .size()
    .reset_index(name='visits')
)

registrations_by_day = (
    registrations
    .assign(date_group=registrations['datetime'].dt.normalize())
    .groupby(['date_group', 'platform'])
    .size()
    .reset_index(name='registrations')
)
```

```python
conversion = visits_by_day.merge(
    registrations_by_day, on=['date_group', 'platform'], how='outer'
)
conversion[['visits', 'registrations']] = (
    conversion[['visits', 'registrations']].fillna(0).astype(int)
)
conversion['conversion'] = conversion['registrations'] / conversion['visits'] * 100
conversion = conversion.sort_values(['date_group', 'platform']).reset_index(drop=True)
conversion.head(10)
```

```python
conversion.describe()
```

```python
# Дата сохраняется как Unix-время в миллисекундах (формат из примера задания)
epoch_ms = (conversion['date_group'] - pd.Timestamp(0)) // pd.Timedelta(milliseconds=1)
conversion.assign(date_group=epoch_ms).to_json('./conversion.json')
```

## Рекламные кампании

Читаем `./ads.csv` с затратами на рекламу и присоединяем к визитам и регистрациям по дате. Для дней без рекламы `cost = 0`, `utm_campaign = 'none'`.

```python
ads = pd.read_csv('./ads.csv', parse_dates=['date'])
ads.info()
ads.head()
```

```python
ads.describe(include='all')
```

```python
ads_by_day = (
    ads
    .assign(date_group=ads['date'].dt.normalize())
    .groupby('date_group')
    .agg(cost=('cost', 'sum'), utm_campaign=('utm_campaign', 'first'))
    .reset_index()
)

totals_by_day = (
    conversion
    .groupby('date_group')[['visits', 'registrations']]
    .sum()
    .reset_index()
)

ads_conversion = totals_by_day.merge(ads_by_day, on='date_group', how='left')
ads_conversion['cost'] = ads_conversion['cost'].fillna(0).astype(int)
ads_conversion['utm_campaign'] = ads_conversion['utm_campaign'].fillna('none')
ads_conversion = ads_conversion.sort_values('date_group').reset_index(drop=True)
ads_conversion.head(10)
```

```python
ads_conversion.groupby('utm_campaign').agg(
    start=('date_group', 'min'),
    end=('date_group', 'max'),
    days=('date_group', 'count'),
    visits=('visits', 'sum'),
    registrations=('registrations', 'sum'),
    cost=('cost', 'sum'),
).sort_values('start')
```

```python
epoch_ms = (ads_conversion['date_group'] - pd.Timestamp(0)) // pd.Timedelta(milliseconds=1)
ads_conversion.assign(date_group=epoch_ms).to_json('./ads.json')
```

## Визуализация

Графики сохраняются в PNG в директорию `./charts`.

```python
import matplotlib

matplotlib.use('Agg')  # чтобы ноутбук выполнялся без дисплея через jupyter execute
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import Image, display

%matplotlib inline

CHARTS_DIR = './charts'
os.makedirs(CHARTS_DIR, exist_ok=True)

sns.set_theme(style='whitegrid')
FIGSIZE = (14, 7)
PLATFORMS = ['web', 'android', 'ios']
PLATFORM_COLORS = dict(zip(PLATFORMS, sns.color_palette('Set2', len(PLATFORMS))))

campaigns = (
    ads_conversion.loc[ads_conversion['utm_campaign'] != 'none']
    .groupby('utm_campaign')['date_group'].min()
    .sort_values()
    .index
    .tolist()
)
CAMPAIGN_COLORS = dict(zip(campaigns, sns.color_palette('tab10', len(campaigns))))
CAMPAIGN_COLORS['none'] = 'lightgray'


def format_date_axis(ax):
    """Подписи дат раз в две недели, чтобы ось не превращалась в кашу."""
    ax.xaxis.set_major_locator(mdates.WeekdayLocator(byweekday=mdates.MO, interval=2))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    ax.tick_params(axis='x', rotation=45)
    ax.set_xlim(
        ads_conversion['date_group'].min() - pd.Timedelta(days=1),
        ads_conversion['date_group'].max() + pd.Timedelta(days=1),
    )


def save_chart(fig, name):
    """Сохраняет график в ./charts/<name>.png и закрывает фигуру."""
    path = os.path.join(CHARTS_DIR, f'{name}.png')
    fig.tight_layout()
    fig.savefig(path)
    plt.close(fig)
    display(Image(path))


def daily_bars(ax, data, column, color='skyblue'):
    ax.bar(data['date_group'], data[column], width=0.8, color=color)
    format_date_axis(ax)


def stacked_bars_by_platform(ax, column):
    pivot = conversion.pivot_table(
        index='date_group', columns='platform', values=column, aggfunc='sum', fill_value=0
    )
    bottom = pd.Series(0, index=pivot.index)
    for platform in PLATFORMS:
        ax.bar(pivot.index, pivot[platform], bottom=bottom, width=0.8,
               color=PLATFORM_COLORS[platform], label=platform)
        bottom += pivot[platform]
    ax.legend(title='platform')
    format_date_axis(ax)


def bars_by_campaign(ax, column):
    for campaign, color in CAMPAIGN_COLORS.items():
        part = ads_conversion[ads_conversion['utm_campaign'] == campaign]
        ax.bar(part['date_group'], part[column], width=0.8, color=color, label=campaign)
    ax.legend(title='utm_campaign')
    format_date_axis(ax)
```

### Конверсия по каждой платформе

Конверсия web на порядок ниже, чем у приложений, поэтому у каждой платформы свой график и своя шкала.

```python
fig, axes = plt.subplots(len(PLATFORMS), 1, figsize=(14, 10), sharex=True)
for ax, platform in zip(axes, PLATFORMS):
    part = conversion[conversion['platform'] == platform]
    mean = part['registrations'].sum() / part['visits'].sum() * 100
    ax.plot(part['date_group'], part['conversion'], color=PLATFORM_COLORS[platform], label=platform)
    ax.axhline(mean, color='black', linestyle='--', linewidth=1, label=f'за период: {mean:.1f}%')
    ax.set(ylabel='conversion, %')
    ax.legend(loc='upper right')
    format_date_axis(ax)
axes[0].set_title('Конверсия по платформам')
axes[-1].set_xlabel('date_group')
save_chart(fig, 'conversion_by_platform')
```

## Эффективность рекламных кампаний

Базовый уровень — средние визиты и регистрации в день без рекламы. Прирост кампании — сколько визитов и регистраций в день она дала сверх базового уровня. Стоимость дополнительной регистрации = затраты в день / прирост регистраций в день.

```python
baseline = ads_conversion.loc[ads_conversion['utm_campaign'] == 'none', ['visits', 'registrations']].mean()

campaign_summary = (
    ads_conversion
    .groupby('utm_campaign')
    .agg(
        start=('date_group', 'min'),
        end=('date_group', 'max'),
        days=('date_group', 'count'),
        cost=('cost', 'sum'),
        visits_per_day=('visits', 'mean'),
        registrations_per_day=('registrations', 'mean'),
        cost_per_day=('cost', 'mean'),
        visits_total=('visits', 'sum'),
        registrations_total=('registrations', 'sum'),
    )
    .sort_values('start')
)
campaign_summary['conversion'] = (
    campaign_summary['registrations_total'] / campaign_summary['visits_total'] * 100
)
campaign_summary['extra_visits_per_day'] = campaign_summary['visits_per_day'] - baseline['visits']
campaign_summary['extra_registrations_per_day'] = (
    campaign_summary['registrations_per_day'] - baseline['registrations']
)
campaign_summary['cost_per_extra_registration'] = (
    campaign_summary['cost_per_day'] / campaign_summary['extra_registrations_per_day']
).where(campaign_summary.index != 'none')

numeric_columns = campaign_summary.select_dtypes('number').columns
campaign_summary[numeric_columns] = campaign_summary[numeric_columns].round(1)
campaign_summary.drop(columns=['visits_total', 'registrations_total'])
```

```python
conversion_by_campaign = (
    conversion
    .merge(ads_conversion[['date_group', 'utm_campaign']], on='date_group')
    .groupby(['utm_campaign', 'platform'])[['visits', 'registrations']]
    .sum()
)
conversion_by_campaign['conversion'] = (
    conversion_by_campaign['registrations'] / conversion_by_campaign['visits'] * 100
)
conversion_by_campaign['conversion'].unstack().loc[campaign_summary.index].round(1)
```
